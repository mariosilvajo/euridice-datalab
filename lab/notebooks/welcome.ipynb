## Variables and calculations

Python remembers values you store in *variables*. Here we compute how many learning hours one ECTS credit represents:

```python
hours_per_ects = 26   # 1 ECTS ≈ 25–28 hours; we use the midpoint
credits = 1
total_hours = hours_per_ects * credits
print(f"{credits} ECTS ≈ {total_hours} learning hours")
```

**Your turn:** change `credits` to `5` in the cell above and re-run it (that is the size of EURIDICE's *Digital Transition for Professionals* course).

## Lists and loops

Data science is about working with *collections* of values, not single numbers:

```python
partner_countries = ["Iceland", "Austria", "Germany", "Portugal", "Netherlands"]

for country in partner_countries:
    print(f"EURIDICE has partners in {country}")
```

## Your first chart

The `matplotlib` library draws charts. Watch how four lines of code become a figure (the first import takes a few seconds while the library loads):

```python
import matplotlib.pyplot as plt

courses = ["AI Ethics", "Applied AI & ML", "Digital Transition"]
ects = [2, 1, 5]

plt.bar(courses, ects, color="#1d3fbf")
plt.ylabel("ECTS credits")
plt.title("EURIDICE microcredential courses")
plt.show()
```
